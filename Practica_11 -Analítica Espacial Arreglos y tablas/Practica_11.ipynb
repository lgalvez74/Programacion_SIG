```python
#EJERCICIO_1
# 1. Importar librerías

import pandas as pd
import numpy as np


# 2. Crear DataFrame de censos

df = pd.DataFrame({
    "Ciudad": ["Bogotá", "Medellín", "Cali", "Barranquilla"],
    "Latitud": [4.7110, 6.2442, 3.4516, 10.9685],
    "Longitud": [-74.0721, -75.5812, -76.5320, -74.7813]
})

print("DataFrame original:\n")
print(df)


# 3. Función vectorizada Haversine

def calcular_distancias(df):
    
    # --------------------------------------------------------
    # Coordenadas de Bogotá
    # --------------------------------------------------------
    
    lat_bogota = 4.7110
    lon_bogota = -74.0721
    
    # --------------------------------------------------------
    # Conversión de grados a radianes
    # --------------------------------------------------------
    
    lat1 = np.radians(lat_bogota)
    lon1 = np.radians(lon_bogota)
    
    lat2 = np.radians(df["Latitud"])
    lon2 = np.radians(df["Longitud"])
    
    # --------------------------------------------------------
    # Diferencias angulares
    # --------------------------------------------------------
    
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    
    # --------------------------------------------------------
    # Fórmula de Haversine vectorizada
    # --------------------------------------------------------
    
    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )
    
    c = 2 * np.arcsin(np.sqrt(a))
    
    # Radio promedio de la Tierra en kilómetros
    R = 6371
    
    # Distancia final
    distancia = R * c
    
    # --------------------------------------------------------
    # Crear nueva columna
    # --------------------------------------------------------
    
    df["Distancia_Bogota"] = distancia
    
    return df


# 4. Ejecutar función

df = calcular_distancias(df)

print("\nDataFrame con distancias:\n")
print(df)
```

```python
#EJERCICIO_1
# Importar librería

import pandas as pd
import numpy as np


# 1. Creación del DataFrame

datos = {
    "Estación": ["Páramo", "Valle", "Costa", "Selva", "Sabana"],
    "Elevación": [3200, 1500, 15, 200, 2600],
    "Precipitación": [850.5, 1200.0, np.nan, 3000.5, np.nan]
}

df = pd.DataFrame(datos)

print("DataFrame original:\n")
print(df)

# 2. Imputación de datos faltantes

# Calcular promedio de precipitaciones válidas
promedio_precipitacion = df["Precipitación"].mean()

# Rellenar valores nulos con la media
df["Precipitación"] = df["Precipitación"].fillna(promedio_precipitacion)

print("\nDataFrame después de imputar valores nulos:\n")
print(df)


# 3. Filtro espacial (Slicing)

# Seleccionar estaciones con elevación > 1000 m
estaciones_altas = df[df["Elevación"] > 1000]

print("\nEstaciones altas:\n")
print(estaciones_altas)


# 4. Cálculo final

precipitacion_promedio_altas = estaciones_altas["Precipitación"].mean()

print("\nPrecipitación promedio anual en estaciones altas:")
print(precipitacion_promedio_altas)
```
